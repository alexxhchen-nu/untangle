## 7. Xunzhilu 墓葬照片数据集 ✅ 已聚类

| 项目 | 详情 |
|------|------|
| 规模 | 543 张图片，228 条墓葬记录 |
| 命名 | `rec-XXXX_YY_hash.jpg`（XXXX=记录号，YY=图片序号）|
| 图片尺寸 | 平均 411×492px，54KB，竖版 JPEG |
| 标签 | ❌ 无元数据（无 CSV/JSON 标注文件）|
| 来源 | R2 `literature` 桶 `xunzhilu-v2/` |

**价值**: ⭐⭐⭐⭐ — 最贴合墓葬研究的真实田野照片，但需要人工标注。

**聚类结果** (ResNet18 特征 + KMeans k=8):
- Cluster 0: 91 imgs / 89 records — 单图记录为主
- Cluster 1: 38 imgs / 16 records
- Cluster 2: 136 imgs / 37 records — 最大多图记录群
- Cluster 3: 55 imgs / 7 records — 集中，可能同类墓葬
- Cluster 4: 77 imgs / 67 records — 分散
- Cluster 5: 33 imgs / 2 records — 极集中 (rec-0019, rec-0077)
- Cluster 6: 35 imgs / 2 records — 极集中 (rec-0055, rec-0087)
- Cluster 7: 78 imgs / 8 records

**关键发现**: xunzhilu-xushui (317 imgs) 是 xunzhilu-v2 的**完全子集**，所有文件相同。
