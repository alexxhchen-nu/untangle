# 中国考古文物朝代预测 — 数据集调研

目标：寻找可用于训练**朝代预测模型**的中国考古文物图像数据集。

调研时间：2026-07-10

## 1. Bronze Ding 青铜鼎数据集 (CVPR 2023)

| 项目 | 详情 |
|------|------|
| 规模 | 4,000+ 张青铜鼎图像 |
| 标签 | **4个朝代 + 11个时期**（细粒度年代标签）|
| 来源 | 5部考古出版物 + 博物馆 |
| 获取方式 | 填写申请表发送至 [email removed] |
| GitHub | https://github.com/zhourixin/bronze-Ding |
| 论文 | Multi-Granularity Archaeological Dating of Chinese Bronze Dings (CVPR 2023) |

**价值**：⭐⭐⭐⭐⭐ — 有朝代标注，专为年代预测设计，学术认可度高。

状态：⏳ 待申请

## 2. Chinese Eaves Tiles 瓦当数据集 (Nature Heritage Science, 2025)

| 项目 | 详情 |
|------|------|
| 规模 | **16,312 张**瓦当图像 |
| 标签 | **8个朝代**（商周 → 明清）|
| 方法 | 田野实拍，用轻量 EfficientNet 做朝代分类 |
| 获取方式 | 查阅论文 Supplementary 或联系作者 |
| 论文 | https://www.nature.com/articles/s40494-025-02049-3 |

**价值**：⭐⭐⭐⭐⭐ — 规模大、朝代覆盖广，瓦当是常见考古断代标型器。

状态：⏳ 待联系作者

## 3. Chinese Ceramics 15 Dynasties (Sensors, 2025)

| 项目 | 详情 |
|------|------|
| 规模 | 覆盖 **15个朝代**，4种釉色，15种器型 |
| 标签 | 朝代 + 釉色 + 器型（多任务）|
| 来源 | 四川大学研究团队 |
| 获取方式 | 查阅论文 Data Availability Statement 或联系作者 |
| 论文 | https://www.mdpi.com/1424-8220/26/4/1312 |

**价值**：⭐⭐⭐⭐⭐ — 朝代覆盖最广，多标签可用于迁移学习。

状态：⏳ 待联系作者
