### ML 训练数据策略（基于资源清单）

**可直接使用的数据**:
1. `xunzhilu-v2` 543 张墓葬照片 — 已聚类为 8 组，等用户标注后可训练

3. `met/all_objectIDs.json` — Met Museum 47万+ 藏品，可按部门筛选中国文物

**Pipeline 构想: PDF → 器物类别 → 搜图 → 训练集**
1. 批量 OCR 跑 228+ 份 PDF → 提取器物清单
2. NLP 归类："泥质灰陶罐" → "陶罐"，"铜鼎" → "青铜鼎"
3. 每个类别搜图：
   - Roboflow MCP 搜索考古数据集
   - Met Museum API 按朝代+类型筛选
   - Google Images / 学术论文图版
4. 按 `朝代/器物类型` 整理 → 训练分类器

**优势**: 数据来源无限（报告自己告诉你该训练什么类别）
**风险**: 考古现场照 vs 博物馆展厅照风格差异大，混训可能影响精度

## 13. 墓葬数据库设计（2026-07-11）

### 设计目标

1. **支持 OCR 提取数据导入** — 从 228+ PDF 报告中批量提取墓葬信息
2. **GIS 可视化** — 为 Track A（数据库+GIS）提供空间查询支持
3. **ML 训练数据关联** — 为 Track B（图像分类）建立器物-图片-数据集映射
4. **人工审核工作流** — 支持标注、修正、质量追踪

### 核心表结构

```
reports (报告)
  └── tombs (墓葬)
        └── artifacts (器物)
              └── images (图片)

sites (遗址)
  └── tombs (墓葬)
```

| 表名 | 记录数 | 用途 |
|------|--------|------|
| `reports` | 1 | 考古报告来源 |
| `sites` | 1 | 遗址/发掘地点（含 GIS 坐标） |
| `tombs` | 2 | 墓葬信息 |
| `artifacts` | 5 | 随葬器物 |
| `material_categories` | 9 | 材质分类枚举 |
| `vessel_types` | 47 | 器型分类枚举 |
| `images` | 0 | 图片资源 |
| `ml_datasets` | 0 | ML 数据集 |
| `extraction_logs` | 0 | OCR 提取日志 |
| `annotations` | 0 | 人工标注/修正 |

### 视图

- `tomb_overview` — 墓葬概览（用于 GIS 可视化）
- `artifact_stats` — 器物统计（用于分析）
- `report_progress` — 报告处理进度

### GIS 支持

```sql
-- GeoJSON 导出示例
SELECT json_object(
    'type', 'Feature',
    'geometry', json_object(
        'type', 'Point',
        'coordinates', json_array(longitude, latitude)
    ),
    'properties', json_object(
        'tomb_number', tomb_number,
        'dynasty', dynasty,
        'site_name', site_name
    )
) AS geojson
FROM tomb_overview
WHERE latitude IS NOT NULL;
```

### 与 CLAUDE.md 提取字段的映射

| CLAUDE.md 字段 | 数据库表.字段 | 示例 |
